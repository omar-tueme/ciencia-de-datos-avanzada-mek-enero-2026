Las siguientes lineas de código fueron escritas con apoyo por parte de ChatGPT

```python
"""
NOT USED -- TO CHANGE COLUMN NAMES BUT IT MAKES SUBSEQUENT CLEANING VERY DIFFICULT

rename_dict = {
    var: info["Question"]
    for var, info in dictionary_non_id_variables.items()
    if var in df_clean1.columns
}

df_clean1 = df_clean1.rename(columns=rename_dict)
"""
```

Se reconoce que es imposible saber, en una escala del 0 al 1, que tan seguido es "algunos días?" por lo que se optó por espaciar las respuestas de manera igual. En este caso, al tener 3 respuestas, el valor se irá aumentando en 0.5 (0.0 --> 0.5 --> 1)

Se seguirá una lógica similar para futuras transformaciónes

Se harán 3 agrupaciones

* "No responde" y "No conoce el cigarro electrónico" se le asignará la etiqueta de no fuma actualmente dado que esta respuesta es la moda

* "algunos días?" se le asignará 0.5 y "todos los días?" se sel asignará 1.0

* Para finalizar, ambas columnas se colapsaran a 1 sola, si una persona fuma o consume cigarro electronico se considerará el valor mayor de fumador (e.g. si salé con 0.5 y 1, se le asignará 1). Si no consume ninguno de los dos, se considerará un no fumador
