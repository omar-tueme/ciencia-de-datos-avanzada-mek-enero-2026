## Limpieza de la base de datos

Empezaremos categorizar todos los " " como nulls

Vamos a aprovechar para remover las columnas de folio ya que no sirven para el análisis

```python
df_clean = df_clean.replace(' ', np.nan)
```

```python
df_clean = df_clean.drop(columns=['FOLIO_I', 'FOLIO_INT'])

df_clean = df_clean.astype('Int64')
```

```python
df_clean.info()
```

Posterior, se van a remover todas las observaciones que no tienen un valor en nuestra variable objetivo. Esto porque dichas observaciones no estarían aportando a este análisis en especifico. Esto la vamos a guardar en un nuevo DataFrame llamado ***df_clean1***

```python
df_clean1 = df_clean.dropna(subset=['d0415']).reset_index(drop=True)

df_clean1.info()
```

Por último, para que el proceso de limpieza sea mucho más sencillo se va a sustituir los valores de las celdas y los títulos de las columnas con lo que se tiene en el diccionario. Esto es meramente transitorio para poder hacer una limpieza de datos más sencilla

```python
"""
NOT USED -- TO CHANGE COLUMN NAMES BUT IT MAKES SUBSEQUENT CLEANING VERY DIFFICULT

rename_dict = {
    var: info["Question"]
    for var, info in dictionary_non_id_variables.items()
    if var in df_clean1.columns
}

df_clean1 = df_clean1.rename(columns=rename_dict)
"""
```

```python
for var, info in dictionary_non_id_variables.items():
    if var in df_clean1.columns:
        mapping = {k: v for k, v in info.items() if isinstance(k, int)}

        df_clean1[var] = df_clean1[var].map(mapping)
```

### Variables de drogas - Alcohol

Recordemos como se distribuye esta variable. Dado que hay riqueza adicional en esta respuesta respecto a la frecuencia de consumo, no es de nuestro interes convertirla en una variable dictómica. Lo que se propone hacer es volvera una variable continua entre 0 y 1 en donde 0 es consumir alcohol 0 días al año y 1 es consumir alcohol 365 días al año. De este modo, se tendrán los siguientes valores =

* Diario = 1

* Semanal = 52 / 365 = 0.142466

* Mensual = 12 / 365 = 0.032977

* Anual = 1 / 365 = 0.002740

* No he consumido en los últimos 12 meses + No responde + Nunca he consumido = 0

```python
def scale_alcohol(row):
    x = row['d0108']
    if x == 'Diario':
        return 1
    elif x == 'Semanal':
        return 52/365
    elif x == 'Mensual':
        return 12/365
    elif x == 'Anual':
        return 1/365
    else:
        return 0

df_final['Alcohol'] = df_clean1.apply(scale_alcohol, axis=1)

df_final['Alcohol'].value_counts()
```
